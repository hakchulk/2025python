### Clustering, 군집화

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
fruits = np.load('data/fruits_300.npy') # 과일 이미지 데이터
```

```python
fruits.shape # 100 * 100 평면이 300개
```

```python
fruits[0,0,:]
```

```python
# fruits 데이터의 첫 번째 이미지를 화면에 출력
plt.imshow(fruits[0], cmap='gray_r') # gray_r은 반전된 회색조
# plt.axis('off') # 축 숨기기
plt.show()
```

```python
fig, axs = plt.subplots(2,3)

axs[0,0].imshow(fruits[0], cmap='gray_r')
axs[0,1].imshow(fruits[100], cmap='gray_r')
axs[0,2].imshow(fruits[200], cmap='gray_r')
axs[1,0].imshow(fruits[50], cmap='gray_r')
axs[1,1].imshow(fruits[150], cmap='gray_r')
axs[1,2].imshow(fruits[250], cmap='gray_r')
fig.suptitle('Fruit image')
plt.show()
```

```python
apple = fruits[0:100].reshape(100, 100 * 100) # 가로 100, 세로 100이었던 2차원 픽셀 배열을 한 줄로 길게(1차원) 늘어뜨린다는 뜻
print(fruits[0:100].shape)
apple.shape
```

```python
pineapple = fruits[100:200].reshape(-1, 100 * 100) # -1을 넣으면 "나머지 차원을 보고 개수는 네가 알아서 계산해"
banana = fruits[200:300].reshape(-1, 100 * 100)     # 여기서는 총 100만 개의 픽셀을 100*100으로 나누니 자동으로 100이 할당됩니다.
```

```python
apple_mean = np.mean(apple, axis=1)
apple_mean
```

```python
# 각 샘플(이미지 1장)의 픽셀 평균값 계산 (axis=1)
apple_mean = np.mean(apple, axis=1)
pineapple_mean = np.mean(pineapple, axis=1)
banana_mean = np.mean(banana, axis=1)

# 히스토그램 그리기
# plt.figure(figsize=(10, 6))
plt.hist(apple_mean, label='apple', alpha=0.5)
plt.hist(pineapple_mean, label='pineapple', alpha=0.5)
plt.hist(banana_mean, label='banana', alpha=0.5)


plt.xlabel('Average Pixel Value')
plt.ylabel('Frequency')
plt.title('Histogram of Pixel Means by Fruit Type')
plt.legend(['apple', 'pineapple', 'banana'])
plt.show()
```

```python
apple_mean = np.mean(apple, axis=0).reshape(100,100)
pineapple_mean = np.mean(pineapple, axis=0).reshape(100,100)
banana_mean = np.mean(banana, axis=0).reshape(100,100)
```

```python
fig, axs = plt.subplots(1,3, figsize=(20,5))
axs[0].imshow(apple_mean, cmap='gray')
axs[1].imshow(pineapple_mean, cmap='gray_r')
axs[2].imshow(banana_mean, cmap='gray_r')
plt.show()
```

```python
abs_diff = np.abs(fruits - apple_mean)
print(abs_diff.shape)
```

```python
abs_mean = np.mean(abs_diff, axis=(1,2))
print(abs_mean.shape)
```

```python
apple_index = np.argsort(abs_mean) # value 의 최소값으로 정렬
apple_index[0:100]
```

```python
fig, axs = plt.subplots(10,10,figsize=(10,10))
for i in range(10):
    for j in range(10):
        axs[i,j].imshow(fruits[apple_index[i*10+j]], cmap='gray_r')
    
```

```python
abs_diff = np.abs(fruits - banana_mean)
abs_mean = np.mean(abs_diff, axis=(1,2))
banana_index = np.argsort(abs_mean) # value 의 최소값으로 정렬
fig, axs = plt.subplots(10,10,figsize=(10,10))
for i in range(10):
    for j in range(10):
        axs[i,j].imshow(fruits[banana_index[i*10+j]], cmap='gray_r')
        axs[i,j].axis('off')
```
