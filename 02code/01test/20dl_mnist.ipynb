```python
cap = cv2.VideoCapture(0) # 0: 기본 카메라, 1: 외장 카메라
while True:
    ret, frame = cap.read() # ret: 성공 여부, frame: 프레임 이미지
    if not ret:
        break
    cv2.imshow('camera', frame)
    if cv2.waitKey(1) & 0xFF == 27: # ESC 키를 누르면 종료
        break
cap.release()
cv2.destroyAllWindows()
```

```python
cap = cv2.VideoCapture(0) # 0: 기본 카메라, 1: 외장 카메라
while True:
    ret, img = cap.read() # ret: 성공 여부, frame: 프레임 이미지
    if not ret:
        break
    
    #grayRGB = cv2.cvtColor(frame, cv2.COLOR_RGB2GRAY) # 그레이스케일 변환, 
    img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) # 그레이스케일 변환
    _, img_gray1 = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY) # 이진화, THRESH_BINARY_INV
    # 경계선. 적응형 이진화. 블록 크기 15, C값 0
    img_adaptive = cv2.adaptiveThreshold(img_gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 15, 0) 

    cv2.imshow('img', img)
    cv2.imshow('img_gray', img_gray)
    cv2.imshow('cam1', img_gray1)    
    cv2.imshow('img_adaptive', img_adaptive)
    if cv2.waitKey(1) & 0xFF == 27: # ESC 키를 누르면 종료
        break
cap.release()
cv2.destroyAllWindows()
```

```python
img = cv2.imread('img/numbers.jpg')
#img = cv2.imread('img/02.png')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
blur = cv2.GaussianBlur(gray, (9, 9), 0) # 가우시안 블러링. 도장 찍기
_, img_binary = cv2.threshold(blur, 127, 255, cv2.THRESH_BINARY) # 이진화
img_adaptive = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 15, 0) 
img_adaptive_blur = cv2.adaptiveThreshold(blur, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 15, 0) 
cv2.imshow('img', img)
cv2.imshow('gray', gray)
cv2.imshow('blur', blur)
cv2.imshow('img_binary', img_binary)
cv2.imshow('img_adaptive', img_adaptive)
cv2.imshow('img_adaptive_blur', img_adaptive_blur) # 가우시안 블러링 후 적응형 이진화. 이것이 가장 깔끔하게 나옴
cv2.waitKey(0)
cv2.destroyAllWindows()

```
